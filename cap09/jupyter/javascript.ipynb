```python
%%javascript


// Cadastro de compras 
let local = `arroz;cereal;eletrica;banana;dino;farinha`;
function cadastrarCompras(compra) {
    // verifica se há dados
    if (local) {
        if (!local.split(";").includes(compra)) {  // verifica se há valores iguais
            local += `;${compra}`;
        }
    } else {
        local = compra; // como não tem valor para junta usa atribuição '=';
    }
}

// função que exibe as compras em ordem alfabetica

function ordemAlfabetica() {
    if (local) {
        const compras = local.split(";").sort().join("\n"); // as compras já estao em ordem de A-Z!
        element.text(compras);
    } else {
        element.text("Não há compras no momento!...");
    }
}

// função que limpa todos os dados do local
```
